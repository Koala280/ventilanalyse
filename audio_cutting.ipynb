```python
from pydub import AudioSegment
from pydub.silence import split_on_silence
import glob
import os
import os.path

DIR = "splitted"

if not os.path.exists('splitted'):
    os.mkdir('splitted')
else:
    filelist = glob.glob(DIR+"/*")
    for f in filelist:
        os.remove(f)
        
# Load the audio file
audio_file = AudioSegment.from_file("rec/rec2.wav", format="wav")
```

Cut the audio

```python
# Define the minimum silence length in milliseconds
min_silence_len = 150
silence_threshold = -50.6  # adjust this value to change the sensitivity to silence

# Split the audio file into tracks based on silence
tracks = split_on_silence(
    audio_file,
    min_silence_len=min_silence_len,
    silence_thresh=silence_threshold
)

print(len(tracks))
for i, track in enumerate(tracks):
    track.export(f"splitted/rec3_{i}.wav", format="wav")
```
