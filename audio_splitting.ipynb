```python
from pydub import AudioSegment
from pydub.silence import split_on_silence
from glob import glob
import os
import os.path

SPLITTED_DIR = "audios/splitted_audio"

FILEPATH = "audios/filtered_audio/denoised/"
FILENAME = "denoised_rec4.wav"
FILE = FILEPATH + FILENAME



# Load the audio file
audio_file = AudioSegment.from_file(FILE, format="wav")
```

Cut the audio

```python
# Define the minimum silence length in milliseconds
min_silence_len = 2
silence_threshold = -65  # adjust this value to change the sensitivity to silence

# Split the audio file into tracks based on silence
segments = split_on_silence(
    audio_file,
    min_silence_len=min_silence_len,
    silence_thresh=silence_threshold,
    # keep_silence=20
)

print(len(segments))

SEGMENT_PATH = f"{SPLITTED_DIR}/{FILENAME.split('.wav')[0]}_segment"
if not os.path.exists(SEGMENT_PATH):
    os.mkdir(SEGMENT_PATH)
else:
    filelist = glob(SEGMENT_PATH+"/*")
    for f in filelist:
        os.remove(f)

for i, segment in enumerate(segments):
    # Speichere die Datei nur, wenn sie eine Mindestlänge hat
    #   if len(segment) > min_silence_duration:

    segment.export(
        f"{SEGMENT_PATH}/{FILENAME.split('.wav')[0]}_{i}.wav", format="wav")
    print(f"Segment {i} von Datei {FILENAME} wurde gespeichert.")
```
