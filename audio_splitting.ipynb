```python
from pydub import AudioSegment
from pydub.silence import split_on_silence
import glob
import os
import os.path

PATHS = [
    "audios/splitted_audio",
    "audios/rec",
    "audios/filtered_audio",
    "audios/filtered_audio/fir",
    "audios/filtered_audio/bandstop",
    "audios/filtered_audio/bandpass",
    "audios/filtered_audio/highpass",
    "audios/filtered_audio/lowpass",
]

SPLITTED_DIR = PATHS[0]

REC_PATH = PATHS[1]
REC_NAME = "rec3.wav"
REC = REC_PATH + REC_NAME

if not os.path.exists(SPLITTED_DIR):
    os.mkdir(SPLITTED_DIR)
else:
    filelist = glob.glob(SPLITTED_DIR+"/*")
    for f in filelist:
        os.remove(f)

# Load the audio file
audio_file = AudioSegment.from_file(REC, format="wav")
```

Cut the audio

```python
# Define the minimum silence length in milliseconds
min_silence_len = 100
silence_threshold = -50.6  # adjust this value to change the sensitivity to silence

# Split the audio file into tracks based on silence
segments = split_on_silence(
    audio_file,
    min_silence_len=min_silence_len,
    silence_thresh=silence_threshold,
    # keep_silence=20
)

print(len(segments))

for i, segment in enumerate(segments):
    # Speichere die Datei nur, wenn sie eine Mindestlänge hat
    #   if len(segment) > min_silence_duration:

    SEGMENT_PATH = f"{SPLITTED_DIR}\\{REC_NAME}_segment"
    if not os.path.exists(SEGMENT_PATH):
        os.mkdir(SEGMENT_PATH)

    segment.export(f"{SEGMENT_PATH}\\rec3_{i}.wav", format="wav")
    print(f"Segment {i+1} von Datei {REC_NAME} wurde gespeichert.")
```
